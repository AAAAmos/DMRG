```python
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fnames = './HC_data/63_*.txt'
files = glob.glob(fnames)
files
```

```python
with open(files[1]) as f: 
    lines = f.readlines()
    beta = []
    ee = []
    for l in lines:
        ee_b = []
        
        data = l.split(',')
        beta.append(float(data[0]))
        
        for d in data[1:]:
            ee_b.append(float(d))
        ee.append(ee_b)
```

```python
b += 1

fig, ax = plt.subplots()

bins = np.linspace(-0.5, 25.5, 26)
ax.hist(ee[b], bins=bins)

ax.set_xlim(-1, 26)
ax.set_xlabel('-log(Singular value)')
ax.set_ylabel('number')
ax.set_title(f'6x3. $\\beta=${beta[b]}. Entanglement spectrum')
plt.show()
```
